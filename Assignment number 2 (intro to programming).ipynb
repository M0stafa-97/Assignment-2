Task 1 : Write a function that takes a list of strings and a number K. The function should
return a new list containing only the strings that have a length greater than K.

```python
my_list = input("enter a list of strings separated by commas : ")
my_list = my_list.split(',')
my_list = [i.strip() for i in my_list]
k = int(input("enter a number : "))
print(f"my list = {my_list}")

def lengthOfString(my_list, k):
    new_list = []
    for i in my_list:
        if len(i) > k:
            new_list.append(i)

    return f"new list of length greater than {k} = {new_list}"

print(lengthOfString(my_list, k))
```

Task 2 : Write a function named area() that calculates the area of either a "circle" or a
"rectangle". The function should take the shape name as the first argument,
followed by the necessary dimensions (radius for circle; length and width for
rectangle).

```python
import math

def area(shape=None, r=0, length=0, width=0):
    if shape == "circle":
        Area = math.pi*r**2
        return f"area of circle = {Area}"
    elif shape == "rectangle":
        Area = length*width
        return f"area of rectangle = {Area}"

print(area("rectangle",length=5,width=4))
print(area("circle",r=5))
```

Task 4 : Write a program to find the sum of all numbers in a list. ensure
the program ignores any elements in the list that are not numbers (like strings or
booleans).

```python
my_list = [10,15,"Ahmed", True, 5, False, 20, "Apple"]
sum = 0
for i in my_list:
    if isinstance(i, (int,float)) and not isinstance(i, bool): # because boolean like 0 or 1 True = 1 and False = 0
        sum+=i

print(f"Total sum = {sum}")
```

```python
class Restaurant():
    def __init__(self):
        self.menu = {}
        self.booK_table = {}
        self.orders = []

    def get_menu(self):
        return self.menu

    def get_orders(self):
        return self.orders

    def get_table_reservation(self):
        return self.booK_table

    def add_item_to_menu(self,item,price):
        self.menu[item] = price

    def set_table_reservation(self,customer_name, book_table):
        self.booK_table[customer_name] = book_table

    def set_orders(self, orders):
        self.orders = orders


    def __str__(self):
        return f"table reservation : {self.booK_table} , orders : {self.orders}"
```

```python
class Customer():
    def __init__(self):
        self.name = None
        self.age = None

    def get_name(self):
        return self.name

    def get_age(self):
        return self.age

    def set_name(self,name):
        self.name = name

    def set_age(self,age):
        self.age = age
```

```python
class Bike():
    def __init__(self):
        self.name = None
        self.price = None

    def get_name(self):
        return self.name

    def get_price(self):
        return self.price

    def set_name(self, name):
        self.name = name

    def set_price(self, price):
        self.price = price
```

```python
class MainBikeRental():
    def __init__(self):
        self.bikes = {}
        self.rentals = []

    def get_bikes(self):
        return self.bikes

    def get_rentals(self):
        return self.rentals

    def set_bikes(self, bikes):
        self.bikes = bikes

    def set_rentals(self, rentals):
        self.rentals = rentals

    def requestBike(self,number_of_bikes,customer):
        if customer.get_age() > 6:
            if number_of_bikes <= len(self.bikes.keys()):
                for i in list(self.bikes)[:number_of_bikes]:
                    self.rentals.append(i)
                self.bikes = list(self.bikes)[number_of_bikes:]
            else:
                print("wait 10 mins")
        else:
            print("customer age less than 6")

    def returnBike(self,rental_time,number_of_bikes):
        total_bill_cost = rental_time*40
        for i in self.rentals[:number_of_bikes]:
            self.bikes.append(i)
        self.rentals = self.rentals[number_of_bikes:]
        return f"total bill cost = {total_bill_cost}"

    def add_bike(self,bike):
        self.bikes[bike.name] = bike.price

    def total_cost(self):
        total_cost = 0
        for bike in self.bikes.values():
            total_cost+=bike
        return f"total_cost = {total_cost}"
```
